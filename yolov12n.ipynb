```python
from ultralytics import YOLO

model = YOLO("yolo12n.pt")
```

```python
model.train(
    data="/kaggle/working/000-13/data.yaml",  # Path to the data.yaml file
    epochs=100,   # Number of training epochs
    batch=32,     # Corrected batch size argument (use 'batch' instead of 'batch_size')
    imgsz=544,    # Image size (640x640 for YOLOv8)   
    optimizer="AdamW",
    patience=10
)
```

```python
import matplotlib.pyplot as plt
from ultralytics import YOLO

# Load the trained model
model = YOLO("/kaggle/input/best_v12/pytorch/default/1/best_V12.pt")

# Perform inference on the test dataset
results = model.predict(
    source="/kaggle/input/test-data/2025-05-04 10_47_32 AM.jpg",  # Path to the test dataset
    conf=0.2,  # Confidence threshold for detection
    imgsz=544,  # Image size for inference
)

# Loop through the results and display images with bounding boxes
for result in results:
    # Annotate the image with detected boxes
    annotated_image = result.plot()  # This adds the boxes and labels to the image

    # Display the image using matplotlib
    plt.imshow(annotated_image)
    plt.axis('off')  # Hide axes for better visualization
    plt.show()

    # Optionally, print the bounding boxes, labels, and confidences
    print(f"Detected boxes: {result.boxes.xywh}")  # Box coordinates (xywh)
    print(f"Labels: {result.names}")  # Object labels
    print(f"Confidences: {result.boxes.conf}")  # Detection confidences
```
